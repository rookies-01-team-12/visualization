```python
from wordcloud import WordCloud
from PIL import Image
import numpy as np
import io
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt

matplotlib.rc('font', family = 'Malgun Gothic')
```

```python
df_total = pd.read_csv("data/merged_data_total.csv")
df_back = pd.read_csv("data/merged_data_backend.csv")
df_front = pd.read_csv("data/merged_data_frontend.csv")
```

```python
def create_wordcloud(df, shape_text):
    try:
        # 전체 텍스트 마스크 생성 (최종 모양)
        fig, ax = plt.subplots(figsize=(15, 8))
        ax.text(0.5, 0.5, shape_text, fontsize=100, 
                ha='center', va='center', fontweight='bold', fontfamily='sans-serif')
        ax.axis('off')

        buf = io.BytesIO()
        plt.savefig(buf, format='png', bbox_inches='tight', pad_inches=0, transparent=True)
        buf.seek(0)
        final_mask = np.array(Image.open(buf))
        plt.close()

        text = str(list(df['skill']))
        
        # 프레임 생성
        frames = []
        for i in range(1, len(shape_text) + 1):
            fig, ax = plt.subplots(figsize=(20, 10))
            for j, char in enumerate(shape_text):
                color = 'black' if j < i else 'white'
                ax.text(0.5 + (j - len(shape_text)/2) * 0.3, 0.5, char, 
                        fontsize=400, color=color,
                        ha='center', va='center', fontweight='bold', fontfamily='sans-serif')
            ax.axis('off')

            buf = io.BytesIO()
            plt.savefig(buf, format='png', bbox_inches='tight', pad_inches=0, transparent=True)
            buf.seek(0)
            mask = np.array(Image.open(buf))
            plt.close()

            wordcloud = WordCloud(font_path='malgun.ttf', width=1400, height=1400,
                                 background_color='white', mask=mask,
                                 colormap='Dark2', max_words=250).generate(text)
            
            fig = plt.figure(figsize=(10, 10))
            plt.imshow(wordcloud, interpolation='bilinear')
            plt.axis('off')

            buf = io.BytesIO()
            plt.savefig(buf, format='png', bbox_inches='tight', pad_inches=0)
            buf.seek(0)
            frames.append(Image.open(buf))
            plt.close()

        # GIF 저장
        output_path=f'data/{shape_text}.gif'
        frames[0].save(output_path, format="GIF", save_all=True, 
                       append_images=frames[1:], duration=300, loop=0)
        print(f"GIF saved to {output_path}")

    except Exception as e:
        print(f"Error: {str(e)}")
```

```python
create_wordcloud(df_total, "FULL-STACK")
create_wordcloud(df_back, "BACK-END")
create_wordcloud(df_front, "FRONT-END")
```
